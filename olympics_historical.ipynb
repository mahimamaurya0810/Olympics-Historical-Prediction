```python

import os
import glob
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Dataset folder
DATA_DIR = "datasets"

# Find all CSV files
csv_files = glob.glob(os.path.join(DATA_DIR, "*.csv"))

print("Total CSV files found:", len(csv_files))

if len(csv_files) == 0:
    raise FileNotFoundError(
        "datasets folder nahi mila. Notebook ka folder path check karo."
    )

# Load all datasets
data = {}

for file in csv_files:
    filename = os.path.basename(file)

    data[filename] = pd.read_csv(file, low_memory=False)

    print("\nFile:", filename)
    print("Rows and columns:", data[filename].shape)

print("\nAll datasets loaded successfully!")
```

```python

for filename, df in data.items():
    print("\n" + "=" * 60)
    print("DATASET:", filename)
    print("=" * 60)

    print("\nColumn names:")
    print(df.columns.tolist())

    print("\nFirst 5 rows:")
    display(df.head())

    print("\nMissing values:")
    print(df.isnull().sum())

```

```python

# Dataset sizes
summary = pd.DataFrame([
    {
        "Dataset": filename,
        "Rows": df.shape[0],
        "Columns": df.shape[1],
        "Duplicate Rows": df.duplicated().sum()
    }
    for filename, df in data.items()
])

display(summary)

# Visualize number of records in each dataset
plt.figure(figsize=(11, 5))
plt.bar(
    [name.replace(".csv", "") for name in summary["Dataset"]],
    summary["Rows"]
)
plt.xticks(rotation=45, ha="right")
plt.ylabel("Number of rows")
plt.title("Olympics Dataset Overview")
plt.tight_layout()
plt.show()
```

```python

# Select Olympics medal tally dataset
medal_df = data["Olympic_Games_Medal_Tally.csv"].copy()

print("Dataset shape:", medal_df.shape)
print("\nColumn names:")
print(medal_df.columns.tolist())

print("\nFirst 10 rows:")
display(medal_df.head(10))

print("\nData types:")
print(medal_df.dtypes)

print("\nMissing values:")
print(medal_df.isnull().sum())
```

```python

# Show numeric columns and their distributions
medal_df.hist(figsize=(14, 10), bins=20)
plt.suptitle("Olympics Medal Tally: Data Distribution")
plt.tight_layout()
plt.show()
```

```python

# Work on a copy of the medal tally dataset
medals = medal_df.copy()

# Select Summer Olympics for the 2028 Summer Olympics project
medals = medals[
    medals["edition"].astype(str).str.contains(
        "Summer Olympics", case=False, na=False
    )
].copy()

# Clean data types
for col in ["year", "gold", "silver", "bronze", "total"]:
    medals[col] = pd.to_numeric(medals[col], errors="coerce")

medals = medals.dropna(
    subset=["year", "country", "gold", "silver", "bronze", "total"]
)

medals = medals[medals["total"] >= 0]
medals["year"] = medals["year"].astype(int)

# Remove duplicate country-edition entries if any
medals = medals.drop_duplicates(
    subset=["edition_id", "country_noc"]
)

print("Summer Olympics dataset shape:", medals.shape)
print("Years covered:", medals["year"].min(), "-", medals["year"].max())
print("Countries:", medals["country"].nunique())
print("Missing values:", medals.isnull().sum().sum())

display(medals.head())
```

```python

# Top 10 countries by total historical medals
top_countries = (
    medals.groupby("country")["total"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_countries.sort_values().plot(
    kind="barh", figsize=(10, 6)
)

plt.title("Top 10 Countries by Historical Summer Olympic Medals")
plt.xlabel("Total medals across recorded editions")
plt.ylabel("Country")
plt.tight_layout()
plt.show()

# Historical medals won by India
india = medals[
    medals["country"].str.lower().eq("india")
].sort_values("year")

if not india.empty:
    plt.figure(figsize=(10, 5))
    plt.plot(india["year"], india["total"], marker="o")
    plt.title("India's Historical Summer Olympics Medal Trend")
    plt.xlabel("Year")
    plt.ylabel("Total medals")
    plt.grid(True)
    plt.tight_layout()
    plt.show()
else:
    print("India country name dataset mein verify karna hoga.")
```

```python

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

from sklearn.linear_model import (
    LinearRegression, Ridge, Lasso, PoissonRegressor
)
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import (
    RandomForestRegressor,
    ExtraTreesRegressor,
    GradientBoostingRegressor,
    HistGradientBoostingRegressor,
    AdaBoostRegressor
)
from sklearn.neighbors import KNeighborsRegressor
from sklearn.svm import SVR
from sklearn.neural_network import MLPRegressor

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# Features and target
X = medals[["country", "year"]].copy()
y = medals["total"].astype(float)

# Test on later years, not randomly mixed years
years = sorted(medals["year"].unique())

if len(years) < 5:
    raise ValueError("Model evaluation ke liye aur historical editions chahiye.")

n_test_years = max(1, int(np.ceil(len(years) * 0.2)))
test_years = years[-n_test_years:]

train_mask = ~medals["year"].isin(test_years)
test_mask = medals["year"].isin(test_years)

X_train, X_test = X.loc[train_mask], X.loc[test_mask]
y_train, y_test = y.loc[train_mask], y.loc[test_mask]

print("Training years:", sorted(X_train["year"].unique()))
print("Testing years:", sorted(X_test["year"].unique()))
print("Training records:", len(X_train))
print("Testing records:", len(X_test))

# Preprocessing: encode countries and scale year
preprocessor = ColumnTransformer([
    ("country", OneHotEncoder(handle_unknown="ignore"), ["country"]),
    ("year", StandardScaler(), ["year"])
], sparse_threshold=0)

models = {
    "Linear Regression": LinearRegression(),
    "Ridge Regression": Ridge(alpha=1.0),
    "Lasso Regression": Lasso(alpha=0.1, max_iter=10000),
    "Poisson Regression": PoissonRegressor(alpha=0.1, max_iter=1000),
    "Decision Tree": DecisionTreeRegressor(
        max_depth=12, random_state=42
    ),
    "Random Forest": RandomForestRegressor(
        n_estimators=200, min_samples_leaf=2, random_state=42, n_jobs=-1
    ),
    "Extra Trees": ExtraTreesRegressor(
        n_estimators=200, min_samples_leaf=2, random_state=42, n_jobs=-1
    ),
    "Gradient Boosting": GradientBoostingRegressor(random_state=42),
    "Hist Gradient Boosting": HistGradientBoostingRegressor(
        max_iter=150, random_state=42
    ),
    "AdaBoost": AdaBoostRegressor(
        n_estimators=100, random_state=42
    ),
    "KNN": KNeighborsRegressor(n_neighbors=5, weights="distance"),
    "SVR": SVR(C=10, epsilon=0.1),
    "MLP Neural Network": MLPRegressor(
        hidden_layer_sizes=(64, 32),
        max_iter=1000,
        early_stopping=True,
        random_state=42
    )
}

results = []
trained_models = {}

for name, estimator in models.items():
    try:
        pipe = Pipeline([
            ("preprocessing", preprocessor),
            ("model", estimator)
        ])

        pipe.fit(X_train, y_train)
        pred = np.maximum(pipe.predict(X_test), 0)

        results.append({
            "Algorithm": name,
            "MAE": mean_absolute_error(y_test, pred),
            "RMSE": np.sqrt(mean_squared_error(y_test, pred)),
            "R2 Score": r2_score(y_test, pred)
        })

        trained_models[name] = pipe
        print("Completed:", name)

    except Exception as e:
        print("Failed:", name, "|", str(e))

comparison = pd.DataFrame(results).sort_values("MAE")
display(comparison.round(3))
```

```python

# Lower MAE is better
best_name = comparison.iloc[0]["Algorithm"]
best_model = trained_models[best_name]

print("Best model by test MAE:", best_name)
print("Test MAE:", round(comparison.iloc[0]["MAE"], 2))
print("Test RMSE:", round(comparison.iloc[0]["RMSE"], 2))
print("Test R2:", round(comparison.iloc[0]["R2 Score"], 3))

comparison.set_index("Algorithm")["MAE"].sort_values().plot(
    kind="bar", figsize=(12, 5)
)
plt.title("Comparison of ML Algorithms")
plt.ylabel("Test MAE (lower is better)")
plt.xticks(rotation=55, ha="right")
plt.tight_layout()
plt.show()
```

```python

# Predict for countries seen in historical training data
countries = sorted(X_train["country"].unique())

future_data = pd.DataFrame({
    "country": countries,
    "year": 2028
})

future_data["Predicted_Total_Medals"] = np.maximum(
    best_model.predict(future_data[["country", "year"]]), 0
).round(0).astype(int)

future_data = future_data.sort_values(
    "Predicted_Total_Medals", ascending=False
).reset_index(drop=True)

future_data.insert(0, "Predicted_Rank", range(1, len(future_data) + 1))

display(future_data.head(20))

future_data.to_csv("olympics_2028_predictions.csv", index=False)
print("Predictions saved to olympics_2028_predictions.csv")
```

```python

# IMPROVED OLYMPICS MEDAL PREDICTION MODEL

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.ensemble import (
    ExtraTreesRegressor,
    RandomForestRegressor,
    GradientBoostingRegressor
)
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Use the cleaned Summer Olympics dataset from Step 7
base = medals.copy()

# One row per country/NOC per Olympic edition
actual = (
    base.groupby(["edition_id", "year", "country_noc"], as_index=False)
    .agg(
        gold=("gold", "sum"),
        silver=("silver", "sum"),
        bronze=("bronze", "sum"),
        total=("total", "sum")
    )
)

# Build an edition sequence
editions = (
    actual[["edition_id", "year"]]
    .drop_duplicates()
    .sort_values("year")
    .reset_index(drop=True)
)

editions["edition_order"] = np.arange(len(editions))
print("Olympic editions:", len(editions))

# All historical NOCs, with a name for display
country_names = (
    base.sort_values("year")
    .drop_duplicates("country_noc", keep="last")
    .set_index("country_noc")["country"]
    .to_dict()
)

nocs = sorted(actual["country_noc"].dropna().unique())

# Create a complete country-edition grid.
# Countries with no recorded medals in an edition get zero medals.
grid = pd.MultiIndex.from_product(
    [editions["edition_id"], nocs],
    names=["edition_id", "country_noc"]
).to_frame(index=False)

grid = grid.merge(editions, on="edition_id", how="left")

grid = grid.merge(
    actual[["edition_id", "country_noc",
            "gold", "silver", "bronze", "total"]],
    on=["edition_id", "country_noc"],
    how="left"
)

medal_cols = ["gold", "silver", "bronze", "total"]
grid[medal_cols] = grid[medal_cols].fillna(0)

# Previous edition's medal counts, aligned by edition order
previous = grid[
    ["edition_order", "country_noc",
     "gold", "silver", "bronze", "total"]
].copy()

previous["edition_order"] += 1
previous = previous.rename(columns={
    "gold": "prev_gold",
    "silver": "prev_silver",
    "bronze": "prev_bronze",
    "total": "prev_total"
})

ml_data = grid.merge(
    previous,
    on=["edition_order", "country_noc"],
    how="left"
)

prev_cols = ["prev_gold", "prev_silver", "prev_bronze", "prev_total"]
ml_data[prev_cols] = ml_data[prev_cols].fillna(0)

# Use only past information as features
feature_cols = [
    "country_noc", "year",
    "prev_gold", "prev_silver", "prev_bronze", "prev_total"
]

# Test on the latest 20% of Olympic editions
all_years = sorted(ml_data["year"].unique())
test_start_index = max(1, int(len(all_years) * 0.8))
test_start_year = all_years[test_start_index]

train = ml_data[ml_data["year"] < test_start_year].copy()
test = ml_data[ml_data["year"] >= test_start_year].copy()

X_train2 = train[feature_cols]
y_train2 = train["total"]

X_test2 = test[feature_cols]
y_test2 = test["total"]

preprocessor2 = ColumnTransformer([
    ("noc", OneHotEncoder(handle_unknown="ignore"), ["country_noc"]),
    ("numeric", StandardScaler(),
     ["year", "prev_gold", "prev_silver", "prev_bronze", "prev_total"])
])

models2 = {
    "Ridge Regression": Ridge(alpha=1.0),
    "Random Forest": RandomForestRegressor(
        n_estimators=200, min_samples_leaf=2,
        random_state=42, n_jobs=-1
    ),
    "Extra Trees": ExtraTreesRegressor(
        n_estimators=200, min_samples_leaf=2,
        random_state=42, n_jobs=-1
    ),
    "Gradient Boosting": GradientBoostingRegressor(
        n_estimators=150, random_state=42
    )
}

results2 = []
trained_models2 = {}

for name, estimator in models2.items():
    pipe = Pipeline([
        ("preprocessing", preprocessor2),
        ("model", estimator)
    ])

    pipe.fit(X_train2, y_train2)
    pred = np.maximum(pipe.predict(X_test2), 0)

    results2.append({
        "Algorithm": name,
        "MAE": mean_absolute_error(y_test2, pred),
        "RMSE": np.sqrt(mean_squared_error(y_test2, pred)),
        "R2 Score": r2_score(y_test2, pred)
    })

    trained_models2[name] = pipe
    print("Completed:", name)

comparison2 = pd.DataFrame(results2).sort_values("MAE")
display(comparison2.round(3))

print("Training years:", train["year"].min(), "-", train["year"].max())
print("Testing years:", test["year"].min(), "-", test["year"].max())
```

```python

# Countries/NOCs recorded in the latest historical edition
latest_year = int(editions["year"].max())
latest_edition_id = editions.loc[
    editions["year"].eq(latest_year), "edition_id"
].iloc[0]

latest_data = grid[
    grid["edition_id"].eq(latest_edition_id)
].copy()

future = pd.DataFrame({
    "country_noc": latest_data["country_noc"].values,
    "year": 2028,
    "prev_gold": latest_data["gold"].values,
    "prev_silver": latest_data["silver"].values,
    "prev_bronze": latest_data["bronze"].values,
    "prev_total": latest_data["total"].values
})

# Select the best model on held-out historical years
best_name2 = comparison2.iloc[0]["Algorithm"]
best_model2 = trained_models2[best_name2]

future["country"] = future["country_noc"].map(country_names)
future["Predicted_Total_Medals"] = np.maximum(
    best_model2.predict(future[feature_cols]), 0
).round().astype(int)

future = future.sort_values(
    "Predicted_Total_Medals", ascending=False
).reset_index(drop=True)

future.insert(0, "Predicted_Rank", range(1, len(future) + 1))

display(future.head(20))

future.to_csv("olympics_2028_improved_predictions.csv", index=False)
print("Saved: olympics_2028_improved_predictions.csv")
print("Model selected:", best_name2)
print("Latest historical edition used:", latest_year)
```

```python

# Remove known historical/non-country Olympic team codes
historical_codes = [
    "URS",  # Soviet Union
    "GDR",  # East Germany
    "FRG",  # West Germany
    "EUN",  # Unified Team
    "YUG",  # Yugoslavia
    "TCH",  # Czechoslovakia
    "ROC"   # Russian Olympic Committee designation
]

clean_predictions = future[
    ~future["country_noc"].isin(historical_codes)
].copy()

clean_predictions = clean_predictions.sort_values(
    "Predicted_Total_Medals",
    ascending=False
).reset_index(drop=True)

clean_predictions["Predicted_Rank"] = range(
    1, len(clean_predictions) + 1
)

display(clean_predictions.head(20))

clean_predictions.to_csv(
    "olympics_2028_clean_predictions.csv",
    index=False
)

print("Saved: olympics_2028_clean_predictions.csv")
```
